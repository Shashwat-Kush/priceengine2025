```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import pandas as pd

# Load the dataset from the CSV file
# We use encoding='latin1' to avoid potential errors with special characters.
try:
    df = pd.read_csv('/Users/shashwatkushwaha/Desktop/AcadProj/Acad/MTP/priceengine2025/datasets/DiwaliSales/Diwali Sales Data.csv', encoding='latin1')
except FileNotFoundError:
    print("Error: 'Diwali Sales Data.csv' not found. Make sure the file is in the correct directory.")
    exit()
```

```python
df.drop(['Status', 'unnamed1'], axis=1, inplace=True)
df.dropna(inplace=True)
df['Amount'] = df['Amount'].astype(int)
df.head(20)
```

```python
df[['Age', 'Orders', 'Amount']].describe()
```

Core customers are adulsts with mean age of 35;
On average, 2-3 orders;
Average transaction amount is around 9.5k;

```python
# Analyzing the 'Gender' column
ax = sns.countplot(x='Gender', data=df)
for bars in ax.containers:
    ax.bar_label(bars)
plt.title('Distribution of Sales by Gender')
plt.show()
```

Females are the higher contributors for sales;


```python
# Analyzing sales by 'Age Group'
age_sales = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Age Group', y='Amount', data=age_sales)
plt.title('Total Sales Amount by Age Group')
plt.show()
```

26-35 is the highest contributor of sales; So females between this age group maybe the primary target;

```python
# --- 2. Plotting Sales by Marital Status ---
plt.figure(figsize=(7, 5))
ax = sns.countplot(data=df, x='Marital_Status')
plt.title('Sales Distribution by Marital Status')
# Adding labels (0 = Unmarried, 1 = Married)
plt.xlabel('Marital Status (0 = Unmarried, 1 = Married)')
for bars in ax.containers:
    ax.bar_label(bars)
```

unmarried are making more sales

```python
# --- 3. Plotting Sales by Occupation ---
plt.figure(figsize=(20, 8)) # Wider figure for readability
sales_by_occupation = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(data=sales_by_occupation, x='Occupation', y='Amount')
plt.title('Total Sales Amount by Occupation')
plt.xticks(rotation=45) # Rotate labels to prevent overlap
```

Top Spending Occupations

```python
# --- 4. Plotting Sales by Product Category ---
plt.figure(figsize=(20, 8)) # Wider figure for readability
sales_by_category = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)
sns.barplot(data=sales_by_category, x='Product_Category', y='Amount')
plt.title('Top 10 Product Categories by Sales Amount')
plt.xticks(rotation=45)
```

Food garners almost double the sales than the rest of them

```python
# --- 5. Plotting Sales by State ---
plt.figure(figsize=(20, 8)) # Wider figure for readability
sales_by_state = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)
sns.barplot(data=sales_by_state, x='State', y='Amount')
plt.title('Top 10 States by Sales Amount')
plt.xticks(rotation=45)
```

```python
# Group by Marital_Status and calculate multiple statistics at once
proof_table = df.groupby('Marital_Status').agg(
    Number_of_Orders=('Orders', 'count'),
    Total_Sales_Amount=('Amount', 'sum')
)

# Add readable labels for clarity
proof_table.index = proof_table.index.map({0: 'Unmarried', 1: 'Married'})
print(proof_table)
```

```python
# Combined Analysis: Sales by Age Group and Gender ---
# We create a bar plot that shows average amount for each age group,
# but also splits the bars by gender.

plt.figure(figsize=(12, 7))
ax = sns.barplot(data=df, x='Age Group', y='Amount', hue='Gender')
plt.title('Mean Amount by Age Group and Gender')
plt.xlabel('Age Group')
plt.ylabel('Mean Amount')
mean_amount_table = df.groupby(['Age Group', 'Gender'])['Amount'].mean().round(2).reset_index()
print("----------- Mean Purchase Amount by Age Group and Gender -----------")
print(mean_amount_table)
```

For most age groups (0-17, 18-25, 26-35, 36-45), the average purchase amount for both men and women is quite similar, generally hovering between ₹9,100 and ₹9,800.

A notable exception is in the older age groups. Men in the 51-55 and 55+ brackets have a significantly higher average spend per transaction (over ₹10,300) compared to any other group

While men shop less frequently, older men tend to make higher-value purchases when they do.

```python
# Combined Analysis: Sales by Age Group and Gender ---
# We create a bar plot that shows total sales amount for each age group,
# but also splits the bars by gender.

plt.figure(figsize=(12, 7))
ax = sns.barplot(data=df, x='Age Group', y='Amount', hue='Gender', estimator=sum)
plt.title('Total Sales Amount by Age Group and Gender')
plt.xlabel('Age Group')
plt.ylabel('Total Sales Amount')
total_sales_table = df.groupby(['Age Group', 'Gender'])['Amount'].sum().reset_index()
print("----------- Total Sales Amount by Age Group and Gender -----------")
print(total_sales_table)
```

females contribute significantly more to total sales than males in every single age bracket. The difference is not small; it's substantial across the board.

The 26-35 age group of females is the undisputed champion, bringing in nearly ₹31 million in sales. This is more than double the revenue from the next closest group (Females, 36-45 at ₹15.5 million).

The primary commercial engine for Diwali sales is women, especially those in the 26-35 age group, then 36-45 age group then 18-25 age group
